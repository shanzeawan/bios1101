## Oppgave 5: Body mass and heart rate in mammals; an *allometric* relationship

<!-- Dataset and model is adapted from Noujaim et al., 2004 *From Mouse to Whale: A universal scaling relation for the PR interval of the electrocardiogram of Mammals" -->

In animals, the relationship between body size to the size and shape of body parts and to physiological processes (i.e. metabolic rate, ageing and heart rate) is rarely linear, but usually follows a power function on the form:

$$
Y = aX^b
$$

where $Y$ is the trait or process in question, $X$ is the total body size (in mass or height/length) and $a$ and $b$ are constants.

Some examples:

* Human heads (babies' heads are relatively larger than adults', and become proportionally smaller as the human grows larger)
* Claw of the fiddler crab (larger crabs have relatively larger claws than smaller crabs)
* Lifespan in mammals (the larger the body the longer the life)
* Heart rate in mammals (the smaller the body the higher the heart rate

*(curiously, the two latter phenomena leading to almost all mammals experiencing approximately the same number of heartbeats during a lifetime)*

There is both positive and negative allometry. In positive allometry $b > 0$, and its implication is that, as the animal's body grows larger, the body part (or process) becomes relatively larger (*fiddler crab claws become relatively larger with body size*). In negative allometry $b < 0$, meaning that the trait in question becomes relatively smaller the larger the body growth (*heads become relatively smaller as the body grows*).

In this exercise we will explore and model the relationship between heart rate and body mass in mammals.

In figuring out what type of model might be suitable for our data,
it is often useful to use *semilog* and *log-log* plots.
If a semilog plot - with one axis plotted on a log scale,
shows the data as a straight line,
we're probably dealing with an exponential model ($Y = b^X$).
However, if only the log-log plot - with *both* axes
plotted on a log scale,
puts the data in a straight line,
a power function ($Y = aX^b$) is likely a better fit.

Now make both a semilog plot, and a log-log plot by adding

to help determine the appropriate model (exponential versus power).

**c)**
To model the relationship between heart rate and body mass,
we need to obtain the parameters of the power function $Y = aX^b$.
We cannot use NumPy's `np.polyfit()` on the data directly,
but we can use it when we log-log transform the data:

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

body_mass_data = pd.read_csv("body_mass_heart_rate_mammals.csv")

body_mass = np.array(body_mass_data["Body Mass"])
heart_rate = np.array(body_mass_data["Heart Rate"])

plt.plot(body_mass, heart_rate, marker="o", linestyle="none")
plt.xlabel("Body mass (kg)")
plt.ylabel("Heart rate (bpm)")
plt.show()



# Take the log10 of the data
bm_log = np.log10(body_mass)
hr_log = np.log10(heart_rate)

log_mass = np.log(body_mass)
log_heart = np.log(heart_rate)

parameters = np.polyfit(log_mass, log_heart, 1)

print(parameters)
````

Now use `np.polyfit()` as you would use for linear data,
performing a fit with 1 degree.
Print out the slop and intercept values you obtain.


````python
fit = np.polyfit(log_mass, log_heart, 1)
slope = fit[0]
intercept = fit[1]
print("Slope:", slope)
print("Intercept:", intercept)
````
