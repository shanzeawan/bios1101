## Oppgave 1: Algeinvasjonen

I uke 36 hadde vi en oppgave i analyse av algevekst i en innsjø.

Løsningsforslaget har programmet gjengitt under.

````python
import matplotlib.pyplot as plt
T = 31          # simulasjonstid (dager)
r = 1           # vekstrate (daglig)
A_maks = 10     # maks areal, km2
A0_m2 = 0.018   # startarel, m2
t0 = 1          # startdag, 1. juni
A = []          # liste for å lagre areal over tid
tid = []        # liste for å lagre tidspunkter
A0_km2 = A0_m2 / 1e6
A.append(A0_km2)
tid.append(t0)
t = 1
while t <= T:
    A_ny = (1 + r) * A[t-1]
    A.append(A_ny)
    tid_ny = tid[t-1] + 1
    tid.append(tid_ny)
    t = t + 1
plt.plot(tid, A, label='Vekstkurve')
plt.axhline(A_maks, ls=':', color='k', label='Maks innsjøareal')
plt.axvline(tid[-2], ls=':', color='r', label='Dag innsjø er dekket')
plt.xlabel('Dag')
plt.ylabel('Areal dekket i km2')
plt.title('Algevekst i innsjø')
plt.legend()
plt.show()
````

## Oppgave 2: Logistisk vekst av bakterier med hjelp av funksjoner

I kapittel 4 i læreboken ble følgende program utviklet for å modellere bakteriell logistisk vekst:

````python
import matplotlib.pyplot as plt

N = 750 # number of time steps
r = 0.035 # growth rate
K = 8e9 # carrying capacity

E = []
t = []

# Initial conditions
E.append(10042)
t.append(60) # minutes

# Calculating number of bacteria and time for each time step
n = 1
while n < N:
    E_n = (1 + r * (1 - E[n-1]/K)) * E[n-1]
    E.append(E_n)
    
    t_n = t[n-1] + 1
    t.append(t_n)
    
    n = n + 1

plt.plot(t, E)
plt.xlabel("Time, t (minutes)")
plt.ylabel("Population size, E")
plt.title("Bacterial growth")
plt.show()
````

## Oppgave 3: Logistisk vekst av havørn med funksjoner

I uke 37 hadde vi en oppgave om modellering av en havørn populasjon
med hjelp av en logistisk vekstmodell.

Oppgaven hadde følgende løsning:

````python
import matplotlib.pyplot as plt
P0 = 10  # Startbestand
P = [P0]
r = 0.1  # vekstrate
t = [0]
K = 500  # Bæreevne
år = 1
while år <= 100:
    P.append(P[år-1] + r * P[år-1] * (1 - P[år-1] / K))
    t.append(år)
    år = år + 1
plt.plot(t, P)
plt.xlabel('Tid (år)')
plt.ylabel('Havørnbestand')
plt.title('Logistisk vekstmodell av havørn')
plt.show()
````
